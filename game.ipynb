```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import random
import time

# إنشاء نص ترحيبي
header = widgets.HTML("<h2>تحدي: اضغط على الزر الذهبي لتربح! 🏆</h2>")

# إنشاء الزر الرخم
annoying_button = widgets.Button(
    description='اضغطني لو تقدر!',
    button_style='warning', # لون أصفر/ذهبي
    layout=widgets.Layout(width='150px', margin='100px 0 0 100px')
)

output = widgets.Output()

def on_click_callback(b):
    with output:
        clear_output()
        # قائمة ردود فعل رخمة
        responses = [
            "بطيء جداً! 😂",
            "جرب مرة تانية في العيد..",
            "الماوس بتاعك معلق ولا إيه؟",
            "أنا هناااااا 👻"
        ]
        
        # تغيير مكان الزر بشكل عشوائي تماماً
        new_margin_top = f"{random.randint(0, 300)}px"
        new_margin_left = f"{random.randint(0, 400)}px"
        
        annoying_button.layout.margin = f"{new_margin_top} 0 0 {new_margin_left}"
        annoying_button.description = random.choice(responses)
        
        # تغيير اللون بشكل عشوائي لزيادة التشتيت
        styles = ['success', 'info', 'warning', 'danger']
        annoying_button.button_style = random.choice(styles)

# ربط الزر بالدالة
annoying_button.on_click(on_click_callback)

# عرض اللعبة
display(header, annoying_button, output)
```
